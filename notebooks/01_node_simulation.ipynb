# 01 - Node Simulation

Simple test for `Node` and `EnergyModel`.

```python
import sys
from pathlib import Path

# Make project root importable when running this notebook
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from app.config.settings import Settings
from app.simulation.energy_model import EnergyModel
from app.simulation.node import Node
```

```python
settings = Settings()
energy_model = EnergyModel(settings)

node = Node(
    node_id=1,
    x=10.0,
    y=20.0,
    initial_energy=settings.INITIAL_ENERGY,
    energy_model=energy_model,
)

print(node)
print("Alive:", node.is_alive())
```

```python
# Sense + transmit a few times and watch battery decrease
for step in range(5):
    node.sense()
    node.transmit()
    print(f"step={step + 1} -> {node}")
```

```python
# Sleep mode uses much less energy
before = node.energy
node.sleep()
print(f"After sleep: energy={node.energy:.4f}, used={before - node.energy:.4f}, state={node.state}")

node.wake_up()
print(f"After wake up: state={node.state}")
```

```python
# Keep transmitting until the node dies
while node.is_alive():
    node.sense()
    node.transmit()

print(node)
print("Alive:", node.is_alive())
print("Packets sent:", node.packets_sent)
```
