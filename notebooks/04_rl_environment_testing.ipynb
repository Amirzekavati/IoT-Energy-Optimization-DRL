# 04 - RL Environment Testing

Smoke test for `IoTEnergyEnv` with a random agent.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from app.config.settings import Settings
from app.rl import IoTEnergyEnv
```

```python
settings = Settings()
settings.NUM_NODES = 5
settings.MAX_STEPS = 40

env = IoTEnergyEnv(settings=settings, seed=42)
obs, info = env.reset()
print("obs shape:", obs.shape)
print("action space:", env.action_space)
print("reset summary:", info["summary"])
```

```python
obs, info = env.reset(seed=42)
total_reward = 0.0

for t in range(settings.MAX_STEPS):
    action = env.action_space.sample()
    obs, reward, terminated, truncated, info = env.step(action)
    total_reward += reward
    if terminated or truncated:
        print(f"stopped at t={t}, terminated={terminated}, truncated={truncated}")
        break

print("total reward:", round(total_reward, 3))
print("final summary:", info["summary"])
```
