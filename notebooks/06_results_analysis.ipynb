# 06 - Results Analysis

Compare baseline policies and (optionally) a trained DQN agent.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from app.analytics import (
    compare_policies,
    comparison_dataframe,
    plot_comparison_grid,
    plot_energy_curves,
)
from app.config.settings import Settings
from app.rl import DQNAgent, train_dqn
```

```python
settings = Settings()
settings.NUM_NODES = 4
settings.MAX_STEPS = 40
settings.TRAIN_TIMESTEPS = 2000

train_result = train_dqn(
    settings=settings,
    total_timesteps=settings.TRAIN_TIMESTEPS,
    model_path=Path(settings.MODEL_DIR) / "dqn_eval_demo",
    eval_episodes=1,
    seed=42,
)
agent = train_result["agent"]
```

```python
comparison = compare_policies(
    settings=settings,
    agent=agent,
    seed=42,
    n_episodes=3,
)

df = comparison_dataframe(comparison["metrics"])
df
```

```python
fig1 = plot_comparison_grid(
    comparison["metrics"],
    save_path=ROOT / "assets" / "charts" / "policy_comparison.png",
)
fig2 = plot_energy_curves(
    comparison["histories"],
    save_path=ROOT / "assets" / "charts" / "energy_curves.png",
)
fig1, fig2
```
