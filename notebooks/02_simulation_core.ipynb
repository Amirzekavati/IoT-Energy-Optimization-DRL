# Simulation Core

Run the network with the always-transmit baseline and compare with always-sleep.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from app.config.settings import Settings
from app.simulation import AlwaysSleepScheduler, Simulator
```

```python
settings = Settings()
settings.NUM_NODES = 10
settings.MAX_STEPS = 50

sim_tx = Simulator(settings=settings, seed=42)
sim_tx.run()
print("Always transmit:", sim_tx.summary())

sim_sleep = Simulator(settings=settings, scheduler=AlwaysSleepScheduler(), seed=42)
sim_sleep.run()
print("Always sleep:", sim_sleep.summary())
```

```python
# Energy over time for always-transmit
for row in sim_tx.history[::5]:
    print(
        f"step={row['step']:3d} alive={row['alive']:2d} "
        f"energy={row['total_energy']:.3f} rx={row['packets_received']}"
    )
```
