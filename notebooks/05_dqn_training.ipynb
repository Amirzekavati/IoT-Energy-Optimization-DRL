# 05 - DQN Training

Train a small DQN agent on `IoTEnergyEnv` and evaluate it.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from app.config.settings import Settings
from app.rl import DQNAgent, evaluate_agent, train_dqn
```

```python
settings = Settings()
settings.NUM_NODES = 4
settings.MAX_STEPS = 40
settings.TRAIN_TIMESTEPS = 3000

result = train_dqn(
    settings=settings,
    total_timesteps=settings.TRAIN_TIMESTEPS,
    model_path=Path(settings.MODEL_DIR) / "dqn_iot_energy_demo",
    eval_episodes=3,
    seed=42,
)

print("saved:", result["model_path"])
print("mean reward:", round(result["evaluation"]["mean_reward"], 3))
print("last summary:", result["evaluation"]["last_summary"])
```

```python
agent = DQNAgent.load(
    result["model_path"],
    settings=settings,
    seed=42,
)
evaluation = evaluate_agent(agent, n_episodes=2, seed=123)
print(evaluation)
```
